```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# The modern way to get the exact spread visualization:
fig = make_subplots(rows=2, cols=1, shared_xaxes=True, 
                    subplot_titles=("Line Comparison", "Spread (A - B)"))

fig.add_trace(go.Scatter(x=df.index, y=df['A'], name='A'), row=1, col=1)
fig.add_trace(go.Scatter(x=df.index, y=df['B'], name='B'), row=1, col=1)

spread = df['A'] - df['B']
fig.add_trace(go.Scatter(x=df.index, y=spread, name='Spread', line=dict(color='red')), row=2, col=1)

fig.update_layout(height=600, showlegend=True)
fig.show()
```
